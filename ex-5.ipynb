```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Simple Line Graph")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 3, 5, 7, 9]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Two Lines Graph")
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 5, 3, 7, 9]

plt.plot(
    x,
    y,
    color='green',
    linestyle='dashed',
    linewidth=3,
    marker='o',
    markerfacecolor='blue',
    markersize=10
)

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Customized Plot")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.scatter(x, y, color='red', s=100)

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Scatter Plot")
plt.show()
```

```python
x = np.arange(0, 10)
y = x * x

plt.plot(x, y, 'g*--', linewidth=2, markersize=10)

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("2D Diagram")
plt.show()
```

```python
x = np.arange(0, 10)
y = x * x

plt.subplot(2, 2, 1)
plt.plot(x, y, 'r--')

plt.subplot(2, 2, 2)
plt.plot(x, y, 'g*-')

plt.subplot(2, 2, 3)
plt.plot(x, y, 'bo')

plt.subplot(2, 2, 4)
plt.plot(x, y, 'go')

plt.show()
```

```python
x = np.arange(0, 4 * np.pi, 0.1)
y = np.sin(x)

plt.plot(x, y)

plt.title("Sine Wave Form")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 12, 14, 16, 18]
y2 = [5, 7, 9, 11, 13]

plt.fill_between(x, y1, color='blue', alpha=0.5)
plt.fill_between(x, y2, color='green', alpha=0.5)

plt.title("Area Chart")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 12, 14, 16, 18]
y2 = [2, 4, 6, 8, 10]
y3 = [5, 5, 5, 5, 5]

plt.stackplot(x, y1, y2, y3,
              labels=['Line1', 'Line2', 'Line3'])

plt.legend(loc='upper left')

plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.title("Stack Plot")

plt.show()
```

```python
names = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 6]

plt.bar(names, values, color='orange')

plt.xlabel("Names")
plt.ylabel("Values")
plt.title("Bar Graph")

plt.show()
```

```python
plt.barh(names, values, color='green')

plt.xlabel("Values")
plt.ylabel("Names")
plt.title("Horizontal Bar Graph")

plt.show()
```

```python

data = [2,5,7,8,10,15,18,20,22,25,28,30,35,40]

plt.hist(data, bins=5, color='purple', alpha=0.7)

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram")

plt.show()
```

```python
data = np.random.normal(0, 1, 100)

plt.boxplot(data)

plt.title("Box Plot")

plt.show()
```

```python
activities = ['Eat', 'Sleep', 'Work', 'Play']
slices = [3, 7, 8, 6]
colors = ['red', 'yellow', 'green', 'blue']

plt.pie(
    slices,
    labels=activities,
    colors=colors,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)

plt.title("Pie Chart")

plt.show()
```
